## NER

Let's detect entities such as `person names`, `organizations`, `locations`, `medical codes`, `time expressions`, `quantities`, `monetary values`, `percentages` etc.

Let's implement Regex rules to detetc such entities!

```python
import re
```

## Examples

```python
q1 = "Andre Miller earns $4,500 per month from his job at Cisco Systems; $900 is withheld from this amount each month for taxes. What type of tax is this most likely to be?"
q2 = "Drew Davis earns $ 4,500 per month from his job at Cisco Systems ; $ 900 is withheld from this amount each month for taxes. What type of tax is this most likely to be ?"
q3 = "You want to save up for a new car that costs $35,990. You only have $32,367.78, but you are confident that you can invest that and earn 5.8% compounded monthly. How many years will it take ti save up for the car?"
q4 = "You want to save up for a new car that costs $ 18, 995. You only have $15,901.86 , but you are confident that you can invest that and earn 6.9 % compounded monthly. How many years will it take ti save up for the car? 2 ."

questions =[q1, q2, q3, q4]
```

## Person Names & Organizations

https://stackoverflow.com/questions/57539531/how-to-extract-person-name-using-regular-expression

```python
name_regex = re.compile('[A-Z]{1}[a-z]{1,24} [A-Z]{1}[a-z]{1,24}')
# name_regex = re.compile(r"^([A-Za-z \-]{2,25})+$", re.MULTILINE)
```

```python
for q in questions:
  res = name_regex.findall(q)
  print(res)
```

## Numbers integer/floating

https://stackoverflow.com/questions/26137955/python-regex-for-number-with-or-without-decimals-using-a-dot-or-comma-as-separat

```python
number_regex = re.compile(r"\$*\s*\d+[.,]?[.,]?\s*\d*")
```

```python
for q in questions:
  res = number_regex.findall(q)
  print(res)
```

```python
# import re

# def extract_numbers(string):
#     pattern = r'(\$[\d,.]+|\d+%)(?:[$%])?'
#     numbers = re.findall(pattern, string)
#     return numbers


import re

def extract_numbers(string):
    #pattern = r'(\$[\d,]+(?:\.\d+)?|\d+%)(?:[$%])?'
    # pattern = r'(\$[\d,]+(?:\.\d+)?|\d+(?:\.\d+)?%)'
    pattern = r'(\$[\d,]+(?:\.\d+)?|\d+(?:,\d+)*(?:\.\d+)?%)'
    numbers = re.findall(pattern, string)
    return numbers
```

```python
string = "The total amount is $1,234.56, which is 75% of the total."
numbers = extract_numbers(string)
print(numbers)
```

```python
numbers = extract_numbers(q3)
print(numbers)
```

```python
numbers = extract_numbers(q4)
print(numbers)
```

```python
q5 = "You want to save up for a new car that costs $35,990,111,222.7. You only have $32,367.78, but you are confident that you can invest that and earn 5,909.888% compounded monthly. How many years will it take ti save up for the car?"
numbers = extract_numbers(q5)
print(numbers)
```

```python
import re

def extract_entities(text):
    person_pattern = r'(?:[A-Z][a-z]+\s)+[A-Z][a-z]+'
    organization_pattern = r'(?:[A-Z][a-z]+\s?)+(?:corporation|incorporated|company|limited|corp\.|inc\.|co\.|ltd\.)?\b'
    # person_pattern = r'(?:[A-Z][a-z]+\s)+[A-Z][a-z]+'
    # organization_pattern = r'(?:[A-Z][a-z]+\s?)+(?:corporation|incorporated|company|limited|corp\.|inc\.|co\.|ltd\.)?(?![a-z])'
    # person_pattern = r'(?:[A-Z][a-z]+\s)+[A-Z][a-z]+'
    # organization_pattern = r'(?:[A-Z][a-z]+\s?)+(?:corporation|incorporated|company|limited|corp\.|inc\.|co\.|ltd\.)(?![a-z])\b'
    # person_pattern = r'(?:[A-Z][a-z]+\s)+[A-Z][a-z]+'
    # organization_pattern = r'(?:[A-Z][a-z]+\s?)+(?:corporation|incorporated|company|limited|corp\.|inc\.|co\.|ltd\.)*(?![a-z])\b'

    persons = re.findall(person_pattern, text)
    organizations = re.findall(organization_pattern, text)

    return persons, organizations

# Example usage
q = "Andre Miller earns $4,500 per month from his job at Cisco Systems; $900 is withheld from this amount each month for taxes. What type of tax is this most likely to be?"

persons, organizations = extract_entities(q)

print("Person Names:")
for person in persons:
    print(person)

print("\nOrganization Names:")
for organization in organizations:
    print(organization)

```
