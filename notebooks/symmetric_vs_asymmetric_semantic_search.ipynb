## Imports

## Semantic Search (Symmetric)

https://www.sbert.net/docs/pretrained_models.html#sentence-embedding-models

`The following models have been specifically trained for Semantic Search: Given a question / search query, these models are able to find relevant text passages.`

```python
model_1 = SentenceTransformer('multi-qa-MiniLM-L6-cos-v1')
```

## Input Sequence Length

`Transformer models like BERT / RoBERTa / DistilBERT etc. the runtime and the memory requirement grows quadratic with the input length. This limits transformers to inputs of certain lengths. A common value for BERT & Co. are 512 word pieces, which corresponde to about 300-400 words (for English). Longer texts than this are truncated to the first x word pieces.`

`By default, the provided methods use a limit fo 128 word pieces, longer inputs will be truncated. You can get and set the maximal sequence length like this:`

```python
print("Max Sequence Length:", model_1.max_seq_length)

# Change the length to 512
model_1.max_seq_length = 512

print("Max Sequence Length:", model_1.max_seq_length)
```

## Test 1

```python
query_embedding = model_1.encode('How big is London')
passage_embedding = model_1.encode(['London has 9,787,426 inhabitants at the 2011 census',
                                  'London is known for its finacial district'])

print("Similarity:", util.dot_score(query_embedding, passage_embedding))
```

## Test 2

```python
query_embedding = model_1.encode('What is the United States population?')
passage_embedding = model_1.encode(['How many people live in the United States?',
                                  'How big is the United States?'])

print("Similarity:", util.dot_score(query_embedding, passage_embedding))
```

## MSMARCO Models

`
MS MARCO is a large scale information retrieval corpus that was created based on real user search queries using Bing search engine. The provided models can be used for semantic search, i.e., given keywords / a search phrase / a question, the model will find passages that are relevant for the search query.`

`The training data constist of over 500k examples, while the complete corpus consist of over 8.8 Million passages.
`

```python
model_2 = SentenceTransformer('msmarco-distilroberta-base-v2')
```

```python
print("Max Sequence Length:", model_2.max_seq_length)

# Change the length to 512
model_2.max_seq_length = 512

print("Max Sequence Length:", model_2.max_seq_length)
```

```python
query_embedding = model_2.encode('How big is London')
passage_embedding = model_2.encode(['London has 9,787,426 inhabitants at the 2011 census',
                                  'London is known for its finacial district'])

print("Similarity:", util.cos_sim(query_embedding, passage_embedding))
```
