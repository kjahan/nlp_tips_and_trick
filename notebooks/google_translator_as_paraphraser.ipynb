## Paraphrase a book description

`In a startling disruption of the space-time continuum — and a violation of the basic maxim of publishing that nothing is ever on time — the publication date of the 2nd edition of my book Code: The Hidden Language of Computer Hardware and Software was bumped up by 12 days. That means that the publication date is today and the book might be available from your favorite book vendor. (Or perhaps not quite yet.)`

https://www.charlespetzold.com/blog/2022/08/Code-2nd-Edition-Now-Available.html



```python
def paraphraser(text, interim_lang='es'):
  translation = translator.translate(text, dest=interim_lang)

  translated_text = translation.text

  # Translate back
  translation_ = translator.translate(translated_text, dest='en')

  paraphrased_text = translation_.text

  orig_sents = text.split('.')
  for sent in orig_sents:
    print(sent.strip())

  print ("\n=================\n")

  trans_sents = paraphrased_text.split('.')
  for sent in trans_sents:
    print(sent.strip())

  return paraphrased_text
```

```python
text = 'In a startling disruption of the space-time continuum — and a violation of the basic maxim of publishing that nothing is ever on time — the publication date of the 2nd edition of my book Code: The Hidden Language of Computer Hardware and Software was bumped up by 12 days. That means that the publication date is today and the book might be available from your favorite book vendor. (Or perhaps not quite yet.)'

paraphrased_text = paraphraser(text)
```
