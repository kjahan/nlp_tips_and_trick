## Convert pdfs to texts

```python
from pypdf import PdfReader

import tqdm
```

```python
filename = 'Robert.pdf'
data_path = "/content/data/"
```

```python
doc_texts = {}

full_fn_path = data_path + filename
reader = PdfReader(full_fn_path)
number_of_pages = len(reader.pages)
print("Filename: {}".format(filename))
print("No of pages: {}".format(number_of_pages))


all_text = []
with tqdm.tqdm(total=number_of_pages) as progress:
  for page_num in range(number_of_pages):
      page = reader.pages[page_num]
      text = page.extract_text()
      all_text.append(text)
      progress.update(1)
```

## Translate each page

```python
all_text_translated = []

with tqdm.tqdm(total=len(all_text)) as progress:
  for text_in in all_text:
    translated_text = translator.translate(text_in, dest='fa').text
    all_text_translated.append(translated_text)
    progress.update(1)
```
