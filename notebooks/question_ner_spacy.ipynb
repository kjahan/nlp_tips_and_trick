```python
import spacy

nlp = spacy.load('en_core_web_sm')

sentence = "Apple is looking at buying U.K. startup for $1 billion"

doc = nlp(sentence)

for ent in doc.ents:
	print(ent.text, ent.start_char, ent.end_char, ent.label_)
```

```python
q1 = "Andre Miller earns $4,500 per month from his job at Cisco Systems; $900 is withheld from this amount each month for taxes. What type of tax is this most likely to be?"

doc = nlp(q1)

for ent in doc.ents:
	print(ent.text, ent.start_char, ent.end_char, ent.label_)
```
