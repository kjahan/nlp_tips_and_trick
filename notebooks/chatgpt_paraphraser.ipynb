This is paraphrase suggested by Language tool:

`My PhD in Computer Science focused on designing and implementing machine learning algorithms to model human behavior.`

`Since 2002, I have been developing ML/NLP models and distributed systems in order to solve commercial challenges in online advertising, search/recommendations, cryptography, fraud detection, and NLP.`

```python
Dest_Langs = ['es', 'de', 'fr', 'fa', 'ja', 'ko']
# Dest_Lang = random.sample(Dest_Langs, 1)[0]
Dest_Lang = 'es'
```

```python
def paraphrase(text, translator, dest_lang):
  print("Intermedia lang: {}".format(dest_lang))
  translation_1 = translator.translate(text, dest=dest_lang)

  translated_text_1 = translation_1.text

  # Translate back
  translation_2 = translator.translate(translated_text_1, dest='en')

  translated_text_2 = translation_2.text
  return translated_text_2
```

## Paraphrase a book description

`In a startling disruption of the space-time continuum — and a violation of the basic maxim of publishing that nothing is ever on time — the publication date of the 2nd edition of my book Code: The Hidden Language of Computer Hardware and Software was bumped up by 12 days. That means that the publication date is today and the book might be available from your favorite book vendor. (Or perhaps not quite yet.)`

https://www.charlespetzold.com/blog/2022/08/Code-2nd-Edition-Now-Available.html



```python
def paraphraser(text, translator, interim_lang='es'):
  translation = translator.translate(text, dest=interim_lang)

  translated_text = translation.text

  # Translate back
  translation_ = translator.translate(translated_text, dest='en')

  paraphrased_text = translation_.text

  orig_sents = text.split('.')
  for sent in orig_sents:
    print(sent.strip())

  print ("\n=================\n")

  trans_sents = paraphrased_text.split('.')
  for sent in trans_sents:
    print(sent.strip())

  return paraphrased_text
```

```python
def paraphraser_2(text, translator, n=5):
  translated_text = text
  for _ in range(n):
    interim_lang = random.sample(Dest_Langs, 1)[0]
    print("Interim lang: {}".format(interim_lang))
    translation = translator.translate(translated_text, dest=interim_lang)
    translated_text = translation.text

  # Translate back
  translation_ = translator.translate(translated_text, dest='en')

  paraphrased_text = translation_.text

  orig_sents = text.split('.')
  for sent in orig_sents:
    print(sent.strip())

  print ("\n=================\n")

  trans_sents = paraphrased_text.split('.')
  for sent in trans_sents:
    print(sent.strip())

  return paraphrased_text
```

```python
def paraphraser_3(text, translator, interim_lang='es', dest_lang='en'):
  translation = translator.translate(text, dest=interim_lang)

  translated_text = translation.text

  # Translate back
  translation_ = translator.translate(translated_text, dest=dest_lang)

  paraphrased_text = translation_.text

  orig_sents = text.split('.')
  for sent in orig_sents:
    print(sent.strip())

  print ("\n=================\n")

  trans_sents = paraphrased_text.split('.')
  for sent in trans_sents:
    print(sent.strip())

  return paraphrased_text
```
